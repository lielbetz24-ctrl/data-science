```python
scores = [3, 5, 5, 7, 10]
min = scores [0] #הגדרת המשתנה המינימלי כאיבר הראשון על מנת שתיבדק כל הרשימה מתחילתה ועד סופה
max = scores [0] # הגדרת המשתנה המקסימלי כאיבר הראשון על מנת שתיבדק כל הרשימה מתחילתה ועד סופה
total = 0 #הגדרת משתנה לטובת חישוב הממוצע
max_count = 0 #הגדרת משתנים לחישוב המספר השכיח
mode_value = scores [0] #הגדרת משתנה לחישוב הערך השכיח
for i in scores:
  total +=i #חישובב סכום הממוצע
  if i < min: #אם קיים איבר ברשימה שקטן יותר מהאיבר ההתחלתי - אז המשתנה min יהיה שווה לו
    min = i
  if i > max: #אם קיים איבר גדול יותר מהאיבר ההתחלתי - המתשנה max יהיה שווה לו
    max = i
  current_count = scores.count(i) #התחלת ספירה כמה פעמים כל משתנה מופיע
#אם האיבר מופיע יותר פעמים מהאיבר שקדם לו - הערך שלו נכנס למשתנה שהגדרנו
  if current_count > max_count:
    max_count = current_count
    mode_value = i
print(mode_value)
print(min)
print(max-min) # טווח שווה למינימום פחות מקסימום
avg = total/len(scores)
print (avg)
middle_index = len(scores) // 2 # מציאת אינדקס האמצע (אורך הרשימה חלקי 2 )
median = scores[middle_index] # שליפת הערך שנמצא באינדקס האמצעי והדפסתו
print("Median:", median)
print("Mode:", mode_value)
```

```python
scores = [3, 5, 5, 7, 10]
total = 0
#לולאת for לחישוב הממוצע
for i in scores:
   total+=i
avg = total/len(scores)
#לולאה לחישוב ההפרש בריבוע מהממוצע
squared_sum = 0
for i in scores:
  diff = i - avg # משתנה שמקבל כל הרצה של הלולאה את הפרש האיבר מהממוצע
  squared_sum += diff**2 #השמה וחיבור של ההפרשים למשתנה
variance = squared_sum/len(scores) #חישוב השונות
sd = variance**0.5 #חישוב סטיית התקן
print("the average is  " , avg)
print("the variance is  " , variance)
print("the sd is  " , sd)
```

```python
server_A = [9, 10, 11]
server_B = [5, 10, 15]

##סעיף א' - חישוב הממוצעים
# חישוב לשרת A בלולאה אחת
total_A = 0
for i in server_A:
    total_A += i
avg_A = total_A / len(server_A)

# חישוב לשרת B בלולאה אחת
total_B = 0
for i in server_B:
    total_B += i
avg_B = total_B / len(server_B)
#הדפסת הממוצעים
print("Server A Mean:", avg_A)
print("Server B Mean:", avg_B)

##סעיף ב - חישוב סטיית התקן
#חישוב לשרת A
squared_sum_A = 0
for i in server_A:
  diff_A = i - avg_A # משתנה שמקבל כל הרצה של הלולאה את הפרש האיבר מהממוצע
  squared_sum_A += diff_A**2 #השמה וחיבור של ההפרשים למשתנה
variance_A = squared_sum_A/len(server_A) #חישוב השונות
sd_A = variance_A**0.5 #חישוב סטיית התקן

#חישוב לשרת B
squared_sum_B = 0
for i in server_B:
  diff_B = i - avg_B # משתנה שמקבל כל הרצה של הלולאה את הפרש האיבר מהממוצע
  squared_sum_B += diff_B**2 #השמה וחיבור של ההפרשים למשתנה
variance_B = squared_sum_B/len(server_B) #חישוב השונות
sd_B = variance_B**0.5 #חישוב סטיית התקן

print("Server A Standard Deviation:", sd_A)
print("Server B Standard Deviation:", sd_B)

#סעיף ג - מהו השרת בעל סטיית התקן הקטנה ביותר?
if sd_A < sd_B:
  print("Server A has the smaller standard deviation.")
elif sd_A > sd_B:
  print("Server B has the smaller standard deviation.")
else:
  print("Both servers have the same standard deviation.")
```

```python
data_A = [2, 3, 4, 5, 6]
data_B = [2, 3, 4, 5, 1000]

#חישוב על הרשימה הראשונה
total_A = 0
for i in data_A:
  total_A += i
avg_A = total_A / len(data_A) #חישוב הממוצע
median_A = data_A[len(data_A) // 2] #חישוב החציון ע"י גישה לאיבר הנמצא במקום של אורך הרשימה חלקי2.
print("Mean of data_A:", avg_A)
print("Median of data_A:", median_A)

#חישוב על הרשימה השנייה
total_B = 0
for i in data_B:
  total_B += i
avg_B = total_B / len(data_B) #חישוב הממוצע
median_B = data_B[len(data_B) // 2] #חישוב החציון ע"י גישה לאיבר הנמצא במקום של אורך הרשימה חלקי2.
print("Mean of data_B:", avg_B)
print("Median of data_B:", median_B)

#בדיקה והדפסה האם החציונים שווים
print(median_A == median_B)
```

```python
times_A = [2, 4, 5, 7, 8, 10, 12, 15]
times_B = [2, 4, 5, 7, 8, 10, 12, 150]

# חישוב הטווח - מכיוון שנתון שהרשימות ממוינות, ניתן להשתמש בערך הראשון כמינימום, ובערך האחרון כמקסימום
range_A  = times_A[-1] - times_A[0]
range_B = times_B[-1] - times_B[0]
print("Range of times_A:", range_A)
print("Range of times_B:", range_B)
#חישוב הIQR של כל רשימה
Q1_A = (times_A[0]+times_A[1])/2
Q3_A = (times_A[4]+times_A[5])/2
IQR_A = Q3_A - Q1_A
print("IQR of times_A:", IQR_A)

Q1_B = (times_B[0]+times_B[1])/2
Q3_B = (times_B[4]+times_B[5])/2
IQR_B = Q3_B - Q1_B
print("IQR of times_B:", IQR_B)

```
